```python
python = {"Ali", "Sara", "Hina"}
ai = {"Sara", "Ahmed", "Hina"}

print("Both courses:", python & ai)
print("Either course:", python | ai)
```

```python
sentence = "the cat and the dog and the bird"

words = sentence.split()
counts = {}

for word in words:
    counts[word] = counts.get(word, 0) + 1

print(counts)
```
