```python
accuracy = 0.94
latency_ms = 180

if accuracy >= 0.90 & latency_ms <= 200:
  print("Eligible for deployment")
elif accuracy >= 0.90 & latency_ms > 200:
  print("Accuracy is good, but latency is too high")
elif accuracy >= 0.80 & accuracy < 0.90:
  print("Needs improvement before deployment")
else:
  print("Reject and investigate the model")
```
